```python
from ctypes import CDLL, POINTER, c_double
from pathlib import Path
import numpy as np
# lib = CDLL(
#     "/Users/bakhodir/Documents/Theoretical Astrophysics/TimurAka/libequations.dylib"
# )

lib = CDLL(
    "./libequations.dylib"
)

print("Library loaded successfully!")
```

```python
lib.equations.argtypes = [
    POINTER(c_double),  # varst
    POINTER(c_double),  # diffs
    c_double,           # impact_par
    POINTER(c_double),  # parbh
]

lib.equations.restype = None


def equations(
    varst: np.ndarray,
    impact_par: float,
    parbh: np.ndarray,
) -> np.ndarray:
    """Call the C equations() function and return diffs."""

    varst = np.ascontiguousarray(varst, dtype=np.float64)
    parbh = np.ascontiguousarray(parbh, dtype=np.float64)

    # The output must have the required number of elements.
    diffs = np.zeros_like(varst, dtype=np.float64)

    lib.equations(
        varst.ctypes.data_as(POINTER(c_double)),
        diffs.ctypes.data_as(POINTER(c_double)),
        c_double(impact_par),
        parbh.ctypes.data_as(POINTER(c_double)),
    )

    return diffs
```

```python
varst = np.array([1.0, 2.0, 3.0, 4.0, 5])
parbh = np.array([1., 0.1])
impact_par = 0.11

diffs = equations(varst, impact_par, parbh)

print(diffs)
```

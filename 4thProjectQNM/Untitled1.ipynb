# Introduce the metric

```python
import numpy as np

def f_metric(r, mpar, qpar):

    return 1.0-(2.0*mpar / r)*(1.0-np.tanh(qpar**2 / (2.0*mpar*r)))

def initial_photon_state(alpha, beta, d, inclination):
    
    r = np.sqrt(
        d**2
        + alpha**2
        + beta**2
    )

    theta = np.arccos(
        (
            d * np.cos(inclination)
            + beta * np.sin(inclination)
        )
        / r
    )

    phi = np.arctan2(
        alpha,
        d * np.sin(inclination)
        - beta * np.cos(inclination)
    )

    dr_dlambda = -d / r

    dtheta_dlambda = (
        -np.cos(inclination)
        + (
            d / r**2
        )
        * (
            d * np.cos(inclination)
            + beta * np.sin(inclination)
        )
    ) / np.sqrt(
        r**2
        - (
            d * np.cos(inclination)
            + beta * np.sin(inclination)
        )**2
    )

    dphi_dlambda = (
        -alpha * np.sin(inclination)
    ) / (
        alpha**2
        + (
            d * np.sin(inclination)
            - beta * np.cos(inclination)
        )**2
    )

    return np.array([
        r,
        theta,
        dr_dlambda,
        dtheta_dlambda,
        phi
    ])
```

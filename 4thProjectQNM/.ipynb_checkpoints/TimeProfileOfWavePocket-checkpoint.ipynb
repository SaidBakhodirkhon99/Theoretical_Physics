```python
# === Import libraries ===
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import root_scalar

# === Metric and potential definitions ===

def f(r, k, lam):
    """Metric function f(r) = 1/(1-k) - 2/r + (lam/r)*log(r/lam)"""
    return 1/(1 - k) - 2/r + (lam/r) * np.log(r/lam)

def df_dr(r, k, lam):
    """Derivative of f(r) with respect to r"""
    return 2/r**2 + (lam/r**2) * (1 - np.log(r/lam))

def V_eff(r, l, mu, k, lam):
    """Effective potential for scalar perturbation"""
    fr = f(r, k, lam)
    dfr = df_dr(r, k, lam)
    return fr * ((l*(l + 1))/r**2 + dfr/r + mu**2)

# === Horizon radius finder ===

def horizon_radius(k, lam):
    """Find event horizon: root of f(r)=0"""
    sol = root_scalar(lambda r: f(r, k, lam), bracket=[1e-3, 50], method='brentq')
    return sol.root

# === Parameters ===
mu = 0.1   # scalar field mass
l = 2      # multipole number
k_values = [0.0, 0.05, 0.1]
lam_values = [0.05, 0.1, 0.2]

# === Plot V(r) ===
plt.figure(figsize=(8,5))
for k in k_values:
    for lam in lam_values:
        rH = horizon_radius(k, lam)
        r = np.linspace(rH + 0.01, 20, 1000)
        V = V_eff(r, l, mu, k, lam)
        plt.plot(r, V, label=f'k={k}, λ={lam}')

plt.xlabel('r', fontsize=13)
plt.ylabel('V(r)', fontsize=13)
plt.title('Effective Potential for QNMs in PFDM + Kalb–Ramond Metric', fontsize=14)
plt.legend(fontsize=9)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import root_scalar

# === Metric and potential ===
def f(r, k, lam):
    return 1/(1 - k) - 2/r + (lam/r) * np.log(r/lam)

def df_dr(r, k, lam):
    return 2/r**2 + (lam/r**2) * (1 - np.log(r/lam))

def V_eff(r, l, mu, k, lam):
    fr = f(r, k, lam)
    dfr = df_dr(r, k, lam)
    return fr * ((l*(l + 1))/r**2 + dfr/r + mu**2)

def horizon_radius(k, lam):
    sol = root_scalar(lambda r: f(r, k, lam), bracket=[1e-3, 50], method='brentq')
    return sol.root

# === Parameters ===
mu = 0.1
l = 0
k = 0.05
lam = 0.2
umin, umax, n = 0, 1000, 200
vmin, vmax = 1e-5, 200
h = (umax - umin)/n

# === Initialize grid ===
Y = np.zeros((n+1, n+1))
vc = 4.0
vl = 2.0

# Horizon and tortoise mapping (approximate for simplicity)
rH = horizon_radius(k, lam)
def r_of_rs(rs):  # direct approximation since f is monotonic
    return rs + 1.654# simple shift; in full code use numeric inversion of ∫dr/f

# === Initial Gaussian pulse ===
for j in range(n+1):
    v = vmin + j*h
    Y[0, j] = np.exp(-10 * (v - vc)**2)
    if abs(Y[0, j]) < 1e-64:
        Y[0, j] = 0
    Y[j, j] = 0

# === Time evolution ===
for i in range(1, n+1):
    for j in range(i+1, n+1):
        u = umin + i*h
        v = vmin + j*h
        rs = (v - u)/2
        rr = r_of_rs(rs)
        VS = V_eff(rr, l, mu, k, lam)
        YS, YE, YW = Y[i-1, j-1], Y[i-1, j], Y[i, j-1]
        Y[i, j] = (YE + YW)*(1 - 0.5*h**2*VS/8) - YS

# === Extract waveform at fixed radius rs_obs ===
rs_obs = 5
data_t, data_psi = [], []
for i in range(n+1):
    v = vmin + i*h
    u = v - 2*rs_obs
    iu = int((u - umin)/h)
    iv = int((v - vmin)/h)
    if 0 <= iu < n and 0 <= iv < n:
        t = 0.5*(u + v)
        data_t.append(t)
        data_psi.append(abs(Y[iu, iv]))

# === Plot waveform ===
plt.figure(figsize=(7,6))
plt.plot(data_t, data_psi, color='red')
plt.yscale('log')
plt.xlabel("t", fontsize=14)
plt.ylabel("|$\Phi$|", fontsize=14)
plt.title("$\ell$ = 0, $n$ = 0, $k$ = 0.05, $\lambda=0.1$", fontsize=15)
plt.grid(alpha=0.5)
plt.savefig("TimeProfile.pdf", format='pdf')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import root_scalar

# === Metric and potential ===
def f(r, k, lam):
    return 1/(1 - k) - 2/r + (lam/r) * np.log(r/lam)

def df_dr(r, k, lam):
    return 2/r**2 + (lam/r**2) * (1 - np.log(r/lam))

def V_eff(r, l, mu, k, lam):
    fr = f(r, k, lam)
    dfr = df_dr(r, k, lam)
    return fr * ((l*(l + 1))/r**2 + dfr/r + mu**2)

def horizon_radius(k, lam):
    sol = root_scalar(lambda r: f(r, k, lam), bracket=[1e-3, 50], method='brentq')
    return sol.root

# Tortiose mapping (very approximate)
def r_of_rs(rs):
    return rs + 1.654

# === Time evolution function ===
def evolve(k, lam):
    mu = 0.1
    l = 2
    umin, umax, n = 0, 1000, 500
    vmin, vmax = 1e-5, 200
    h = (umax - umin)/n

    Y = np.zeros((n+1, n+1))
    vc = 4.0
    vl = 2.0

    # Initial Gaussian pulse
    for j in range(n+1):
        v = vmin + j*h
        Y[0, j] = np.exp(-10 * (v - vc)**2)
        if abs(Y[0, j]) < 1e-64:
            Y[0, j] = 0
        Y[j, j] = 0

    # Evolution
    for i in range(1, n+1):
        for j in range(i+1, n+1):
            u = umin + i*h
            v = vmin + j*h
            rs = (v - u)/2
            rr = r_of_rs(rs)
            VS = V_eff(rr, l, mu, k, lam)
            YS, YE, YW = Y[i-1, j-1], Y[i-1, j], Y[i, j-1]
            Y[i, j] = (YE + YW)*(1 - 0.5*h**2*VS/8) - YS

    # Extract waveform
    rs_obs = 5
    data_t, data_psi = [], []
    for i in range(n+1):
        v = vmin + i*h
        u = v - 2*rs_obs
        iu = int((u - umin)/h)
        iv = int((v - vmin)/h)
        if 0 <= iu < n and 0 <= iv < n:
            t = 0.5*(u + v)
            data_t.append(t)
            data_psi.append(abs(Y[iu, iv]))

    return np.array(data_t), np.array(data_psi)


# === Try multiple k values ===
k_values = [0.05, 0.1, 0.2]
lam = 0.1  # Fixed λ

plt.figure(figsize=(7,6))

for k in k_values:
    t, psi = evolve(k, lam)
    plt.plot(t, psi, label=f"k = {k}")

plt.yscale("log")
plt.xlabel("t", fontsize=14)
plt.ylabel("|Φ(t)|", fontsize=14)
plt.title(r"Waveform for $\lambda = 0.1$ and different $k$", fontsize=15)
plt.legend()
plt.grid(alpha=0.5)
plt.tight_layout()
plt.show()
```
